## 문제 3

for, range 사용 1부터 100까지 더하기

```python
total = 0
for i in range(1, 100+1):
  total += i

print(total)
```

## 문제 4

랜덤으로 기온 (학습지 32페이지 7번)

* &gt;=25 여름옷
* &gt;=10 가벼운옷
* &lt;10 겨울옷

```python
import random

temperature = random.randint(-20, 40)

print(f"오늘의 기온은 {temperature}도입니다.")

if temperature >= 25:
  print("여름옷을 입으세요")
elif temperature >= 10:
  print("가벼운 옷을 입으세요")
else:
  print("겨울옷을 입으세요")
```
