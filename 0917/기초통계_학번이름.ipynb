- 데이터를 본격적으로 분석하거나 그래프를 그리기 전에 **기초 통계**를 먼저 확인한다.  
기초 통계를 통해 데이터의 전체적인 특징을 빠르게 파악할 수 있다.  
평균, 중앙값, 최댓값, 최솟값 등을 살펴보며 데이터의 분포와 특성을 이해한다.

---
**📘 이번 단원에서 다루는 내용**
- 데이터 구조 확인 (`info()`, `shape`)
- 기초 통계 (`describe()`, `mean()`, `median()`, `mode()`)
- 범주형 데이터 빈도 (`value_counts()`)
- 정렬 (`sort_values()`)
- 그룹별 비교 (`groupby()`)

### **(0) 데이터프레임 생성**

```python
import pandas as pd

data = {
    '이름': ['김하늘', '이준서', '박지민', '최서윤', '정민호'],
    '성별': ['F', 'M', 'F', 'F', 'M'],
    '키': [162, 178, 155, 160, 182],
    '몸무게': [50, 72, 47, 54, 80],
    '혈액형': ['A', 'B', 'O', 'A', 'AB']
}

df = pd.DataFrame(data)
df
```

### **(1) 데이터 구조 확인 (info, shape)**

```python
# 데이터프레임의 구조, 컬럼 정보, 결측치 개수 확인
df.info()
```

```python
# 데이터프레임의 행(row)과 열(column) 개수 확인
df.shape
```

### **(2) 기초 통계 (describe, mean, median, mode)**

```python
# 수치형 데이터의 기초 통계 정보(개수, 평균, 표준편차, 최소값, 최대값 등) 확인
df.describe()
```

```python
# 평균값 계산
df['키'].mean()
```

```python
# 중앙값 계산
df['몸무게'].median()
```

```python
# 최빈값 확인 - '혈액형' 열에서 가장 많이 나타나는 값 확인
df['혈액형'].mode()
```

### **(3) 범주형 데이터 분포 (value_counts)**

```python
# 특정 칼럼의 값별 빈도 계산 - '혈액형' 열의 값별 개수 확인
df.value_counts('혈액형')
```

### **(4) 정렬(sort_values)**

```python
# 정렬 - '키' 열을 기준으로 오름차순 정렬
df.sort_values(by="키")
```

```python
# 정렬 - '키' 열을 기준으로 내림차순 정렬
df.sort_values(by="키", ascending=False)
```

### **(5) 그룹별 비교(groupby)**

```python
# 그룹별 평균 계산 - '성별'을 기준으로 그룹을 나눈 뒤 '몸무게' 평균 계산
df.groupby('성별')['몸무게'].mean()
```

### **🎯퀴즈**

```python
print("1번 문제 : 성별 빈도를 확인해보자.")
df.value_counts('성별')
```

```python
print("2번 문제 : 혈액형별 평균 키를 구해보자.")
df.groupby('혈액형')['키'].mean()
```

```python
print("3번 문제 : 몸무게가 무거운 순서대로 정렬해보자.")
df.sort_values('몸무게', ascending=False)
```
