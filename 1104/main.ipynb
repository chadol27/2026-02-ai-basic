```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

person1 = Person("홍길동", 30)
print(f"이름: {person1.name}, 나이: {person1.age}")
```

```python
class Animal:
  def __init__(self, name):
    self.name = name

  def speak(self):
    print(self.name, "소리를 냅니다.")

class Dog(Animal):
  def speak(self):
    print(self.name, "멍멍!")

class Cat(Animal):
  def speak(self):
    print(self.name, "야옹!")

dog = Dog("멍멍이")
cat = Cat("순대")

dog.speak()
cat.speak()
```
