🎯 2장 연습 문제에서 다룬 데이터프레임을 다시 사용하여 문제를 해결해 보자.

```python
import pandas as pd

data = {
    '메뉴코드': ['M01','M02','M03','M04','M05'],
    '메뉴명': ['아메리카노','카페라떼','카푸치노','초코라떼','녹차라떼'],
    '카테고리': ['커피','커피','커피','음료','음료'],
    '가격': [3000,4000,4500,5000,4800],
    '판매량': [120,95,80,60,70]
}

df = pd.DataFrame(data)
df = df.set_index('메뉴코드')
df
```

1. 가격의 평균과 중앙값을 구하시오.

① 평균

```python
df['가격'].mean()
```

② 중앙값

```python
df['가격'].median()
```

2. 카테고리의 최빈값을 구하시오.

```python
df['카테고리'].mode()
```

3. 카테고리별 메뉴 개수를 구하시오.

```python
df.value_counts('카테고리')
```

4. 가격을 기준으로 내림차순 정렬하시오.

```python
df.sort_values(by="가격", ascending=False)
```

5 카테고리별 평균 판매량을 구하시오.

```python
df.groupby('카테고리')['판매량'].mean()
```
