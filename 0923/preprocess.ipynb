```python
df2['성별'] = df2['성별'].replace({'남자':'남', '여자':'여'})
df2
```

```python
df = pd.concat([df1, df2])
df
```

```python
df['계약번호'].duplicated().any()
```

```python
df = df.drop_duplicates(subset='계약번호', keep="first")
df
```

```python
df = df.dropna()
df
```
